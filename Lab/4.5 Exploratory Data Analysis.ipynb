# Dataset Description
This dataset provides records for 200,000 properties that were either bought or not bought. The data lists what country, city, what the type of property is, the square footage, the price, year it was constructed, and other information regarding the buyer. Ultimately the target feature is whether or no the house was sold or not which is indicated by a 0 for not sold and 1 for sold. There are multiple types of properties, large and small, old and modern, and located all around the world and major cities. I will list the features with a brief description of the values within the csv.

* Country - Self explanatory but it is the Country where the property is located.
* City - In what city the property is located.
* Property Type - Farmhouse, Apartment, Townhouse, Condo, etc...
* Furnishing Status - Whether or not the property is semi-furnish, fully-furnished, or unfurnished.
* Property Size - The square footage of the property.
* Price - The listing price of the property.
* Constructed Year - When the property was built.
* Previous Owners - How many previous owners there are
* Rooms - How many rooms the property has.
* Customer Salary - Salary of the person that is purchasing the home.
* Loan Amount - The loan amount for the purchase of the home.
* Loan Tenure (Years) - How long the loan was for in years.
* Monthly Expenses - Assuming this is the monthly expense estimation for the property.
* Down Payment - The required down pament for the loan.
* EMI to Income Ratio - Im assuming this is similar to the debt-to-income ratio.
* Satisfaction Score - Assuming this is a scale from 1 to 10 on how satisfied the purchaser is with the property.
* Neighborhood Rating - Assuming this is another scale from 1 to 10 on how good the neighborhood is.
* Connectivity Score - Unsure what this is but also appears to be on a scale from 1 to 10.
* Decision - This is the target of the dataset, either a 0 for not sold or 1 for sold.

My initial thoughts for the data that is most related to the target is price, satisfaction score, and property type. I think the least related to the target would be furnishing status, previous owners, and the year constructed.

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.histplot(x='price', data=df, hue='decision')

plt.show()
```

```python
sns.countplot(x='satisfaction_score', data=df, hue='decision')

plt.show()
```

```python
sns.countplot(x='property_type', data=df, hue='decision')

plt.xticks(rotation= 90, ha='right')
plt.show()
```

```python
sns.countplot(x='furnishing_status', data=df, hue='decision')

plt.show()
```

```python
sns.countplot(x='previous_owners', data=df, hue='decision')

plt.show()
```

```python
sns.histplot(x='constructed_year', data=df, hue='decision')

plt.show()
```
