```python
import pandas as pd
import plotly.express as px

df = pd.read_csv('Valve_Player_Data.csv')
df.head()
```

# The Dataset:
I got this data set off of [Kaggle](https://www.kaggle.com/datasets/jackogozaly/steam-player-data). Upon looking at the data it ranges from July of 2012 to September of 2021 which was the date of the posting on Kaggle. Ironically during this timeset the most popular game every single month was Counter-Strike: Global Offensive.

The data contains the following information:
* Average players by month for the current top 100 games
* The average player count
* The player gain
* The percent gain
* The peak player count

```python
fig1 = px.line(
    df, x="Date", y="Avg_players",
    title="CS:GO Average Players Over Time",
    labels={"Date": "Date", "Avg_players": "Average Players"},
)
fig1.show()
```

# Figure 1:
The average number of CS:GO players shows clear fluctuations across the years, with periods of rapid growth (mid 2017 to early 2018) as well as declines following major update releases (late 2018).

```python
fig2 = px.line(
    df, x="Date", y=["Avg_players", "Peak_Players"],
    title="CS:GO Average vs Peak Players Over Time",
    labels={"value": "Players", "Date": "Date", "variable": "Player Metric"}
)
fig2.show()
```

# Figure 2:
Peak player counts are consistently higher than average players, but both follow similar paths. Peaks rise significantly during surges of populatiry, while averages reflect steadier long-term engagements.

```python
fig3 = px.bar(
    df, x="Date", y="Percent_Gain",
    title="Monthly Percent Change in CS:GO Average Players",
    labels={"Date": "Date", "Percent_Gain": "Percent Change"},
    color="Percent_Gain",
    color_continuous_scale="RdYlGn"
)
fig3.show()
```

# Figure 3:
The percent change visualization highlights volatility in player counts, with some months showing double digit loses while others show quite significant growth during patch releases and competitive seasons. This is true with most games that have a competitive season ladder, usually the first 4-7 weeks are the most popular times while the last 2-3 weeks are the biggest loss of players.
